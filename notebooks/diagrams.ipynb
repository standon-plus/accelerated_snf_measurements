```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from snf.swe import j_index, psi
from snf.sampling import *
```

# Define Setup

```python
N = 10 # AUT has 10 polar modes
radius = 1.0 # Measurement distance = 1 m
freq = 25 * 10**9 # Measurement frequency = 25 GHz
kA = radius * 2 * np.pi * freq / scipy.constants.c

# Construct x-oriented dipole for probe
Np = 1
mus = np.array([-1, 1])
R = np.zeros(6, complex)
R[j_index(2, 1, 1)] = -1
R[j_index(2, -1, 1)] = 1
```

# Set Measurement Points

```python
num_points = 100

# theta, phi, idx = equiangular_bernoulli(num_points, N)
# theta, phi = spiral_random_phi(num_points)
# theta, phi = constant_velocity(num_points)
theta, phi = randomized_constant_velocity(num_points)

th_pos, ph_pos = sphere_position(theta, phi)

plt.scatter(th_pos, ph_pos)
plt.xlabel("Theta angle (rad)")
plt.ylabel("Phi angle (rad)")
```

```python
# Extend angles to include chi
# Chi will be 0, pi/2

chi, theta, phi = add_chi(theta, phi)
```

# Calculate Coherence

```python
Psi, sw = psi(chi, theta, phi, R, N, kA, Np, mus, weighted=True)
print(Psi.shape)
```

```python
mu = coherence(Psi)
welch = welch_bound(*Psi.shape)
print(f'Measured coherence: {mu}. Welch lower bound: {welch}')
```
